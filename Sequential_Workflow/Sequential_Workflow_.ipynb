# Block 2: State Define Karein

```python
from typing import TypedDict
from langgraph.graph import StateGraph, START, END

# Workflow ke darmiyan share hone wala data
class WorkflowState(TypedDict):
    input_text: str
    step1_output: str
    final_output: str

print("State successfully defined!")
```

# Step Functions (Nodes) Banayein

```python
# Step 1: Input text ko uppercase karein
def step_one(state: WorkflowState) -> dict:
    text = state["input_text"].upper()
    print(f"[Step 1 Complete]: {text}")
    return {"step1_output": text}

# Step 2: Formatted text ke sath status add karein
def step_two(state: WorkflowState) -> dict:
    res = f"{state['step1_output']} - [PROCESSED SUCCESSFULLY]"
    print(f"[Step 2 Complete]: {res}")
    return {"final_output": res}
```

# Graph Build, Edges Connect aur Compile Karein

```python
# Graph initialize karein
builder = StateGraph(WorkflowState)

# Nodes add karein
builder.add_node("step_1", step_one)
builder.add_node("step_2", step_two)

# Sequential Edges connect karein: START -> Step 1 -> Step 2 -> END
builder.add_edge(START, "step_1")
builder.add_edge("step_1", "step_2")
builder.add_edge("step_2", END)

# Graph compile karein
graph = builder.compile()
print("Graph successfully compiled!")
```

# Workflow Run (Invoke) Karein

```python
# Input data
initial_input = {"input_text": "hello google colab"}

# Workflow run karein
result = graph.invoke(initial_input)

# Final Result print karein
print("\n--- Final Output ---")
print(result["final_output"])
```
